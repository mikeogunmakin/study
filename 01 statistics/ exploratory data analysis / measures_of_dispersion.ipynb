# **Measures of Dispersion**

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter
import pandas as pd
import numpy as np
```

```python
df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "laotse/credit-risk-dataset",
    "credit_risk_dataset.csv",
)
```

```python
df.head()
```

## Variance

Average squared distance of values from the mean.

**when to use:** Mainly for statistical calculations and modelling.

```python
#using pandas
print('variance:', df['loan_int_rate'].var())

# using np
print('variance:', np.var(df['loan_int_rate']))
```

## Standard Deviation

Typical spread of values around the mean.

**when to use:** For roughly symmetric data with few/no major outliers.

```python
#using pandas
print('std:', df['loan_int_rate'].std())

# using numpy
print('std:', np.std(df['loan_int_rate']))
```

## Interquantile Range (IQR)

Range of the middle 50% of the data (Q3 − Q1).

**when to use:** For skewed data or data containing outliers.

```python
#using pandas
iqr = df['loan_int_rate'].quantile(0.75) - df['loan_int_rate'].quantile(0.25)
print('iqr:', iqr)


#using numpy
# Filter out NaN values before calculating, results in NaN if NaN are present.
loan_int_rate_cleaned = df['loan_int_rate'].dropna()
q75, q25 = np.percentile(loan_int_rate_cleaned, [75 ,25])
iqr = q75 - q25
print('iqr:', iqr)
```

## Median Absolute Deviation (MAD)

Median of the absolute distances from the median.

**when to use:** For data with strong outliers or skewness; very robust to extreme values.

```python
from statsmodels.robust import mad

# using stasmodels
# Filter out NaN values before calculating, results in NaN if NaN are present.
loan_int_rate_cleaned = df['loan_int_rate'].dropna()

print('mad:', mad(loan_int_rate_cleaned))
```
