# **Classification**

Machine learning is the **process whereby computers learn to make decisions from data without being explictedly programmed**.
<br><br>
Example

*   learning to predict whether an email is spam or not.
*   learning to cluster customers based on their engagement and purchasing behaviour.

<br>

Unsupervised Learning is **the process of uncovering hidden patterns and structures from unlabeled data.**
For example, a business may wish to group its customers into distinct categories based on their purchasing behavior without knowing in advance what these categories are. This is known as clustering, one branch of unsupervised learning.

<br>

Supervised learning is a type of machine learning where the values to be predicted are already known, and a model is built with the aim of accurately predicting values of previously unseen data. For example, predicting the species of a flower based on the length and width of its petals.
<br><br>
There are 2 types of supervised learning: Classification and Regression.

Classification is used to predict the label, or category, of an observation. For example, we can predict whether a bank transaction is fraudulent or not. As there are two outcomes here - a fraudulent transaction, or non-fraudulent transaction, this is known as binary classification.
<br><br>
 Regression is used to predict continuous values. For example, a model can use features such as number of bedrooms, and the size of a property, to predict the target variable, price of the property.
