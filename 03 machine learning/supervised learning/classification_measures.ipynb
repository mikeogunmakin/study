# **Classification Measures**

## **Accuracy**

The number of correct predictions out of total predictions (or observations).

$$
\text{Accuracy} = \frac{ \text{Correct Predicitions} }{ \text{Total Predictions} } = \frac{TP + TN}{TP + TN + FP + FN}
$$

## **Precision**

The accuracy of positive predictions. Out of the predictions that are positive how many were correct?

- Accuracy
- Precision
