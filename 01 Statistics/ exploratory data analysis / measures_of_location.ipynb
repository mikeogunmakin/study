# **Measures of Location**

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter
import pandas as pd
import numpy as np
from scipy.stats import trim_mean
```

```python
df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "laotse/credit-risk-dataset",
    "credit_risk_dataset.csv",
)
```

## **Mean**

The sum of all values divided by the number of values.
<br><br>

**when to use:** When all observations are equally important and there are no major outliers

```python
# using pandas
print('Mean:', df['person_age'].mean())

# using numpy
print('Mean:', np.mean(df['person_age']))
```

## **Weighted Mean**

The sum of all values times a weight divided by the sum of the weights.
<br><br>

**when to use:** When observations have different importance, frequency, or weights

```python
df_dummy = pd.DataFrame({
    'value': [10, 20, 30, 40],
    'weight': [1, 2, 3, 4]
})

weighted_mean = np.average(df_dummy['value'], weights=df_dummy['weight'])

print('Weighted mean:', weighted_mean)
```

## **Trimmed Mean**

Trimmed mean is calculated by droping a fixed number of sorted values at each end, and then taking the average of the remaining values.
<br><br>
**when to use:** When you want an average but the data contains some extreme values/outliers

```python
# from scipy.stats, trimming top and bottom 10% (drop 10% from each end)
print('Trimmed Mean:', trim_mean(df['person_age'], 0.1))
```

##**Median**

The middle number in a sorted list of the data.

If there is an even number of data valyes, the middle one is one that is not in the data set, but rather the average of the the 2 values that divided the sorted data into lower and upper halves.
<br><br>
**when to use:** When data is skewed or has strong outliers and you want the typical middle value


```python
# using pandas
print('Median:', df['person_age'].median())

# using numpy
print('Median:', np.median(df['person_age']))
```
